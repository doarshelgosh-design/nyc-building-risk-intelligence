```python
import sys

PROJECT_ROOT = "/workspace/nyc-building-risk"
COMMON_PATH = f"{PROJECT_ROOT}/spark/common"

if COMMON_PATH not in sys.path:
    sys.path.insert(0, COMMON_PATH)

from spark_session import create_spark_session
from minio_config import minio_path

spark = create_spark_session(
    "NYC Building Risk - DOB Silver Validation"
)

print("Spark version:", spark.version)
print("Spark UI:", spark.sparkContext.uiWebUrl)
```

```python
silver_dob_path = minio_path(
    "silver/dob"
)

dob_silver_df = (
    spark.read
    .parquet(silver_dob_path)
)

print(
    "DOB Silver rows:",
    dob_silver_df.count()
)
```

```python
dob_silver_df.groupBy(
    "violation_year",
    "violation_month"
).count().orderBy(
    "violation_year",
    "violation_month"
).show(
    50,
    truncate=False
)
```
